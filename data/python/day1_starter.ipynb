## 9. Put them together (graded as part of HW 5.1)

Expected: card first, then payme. (Duplicates are still inside — pandas removes them on Day 3.)

```python
counts = {}
with open(DATA / "excel" / "payments_dirty.csv", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        m = normalise_method(row["method"])
        counts[m] = counts.get(m, 0) + 1
print(counts)
print(top_n(counts, 3))
assert sum(counts.values()) == 52_875 and len(counts) == 5
```
